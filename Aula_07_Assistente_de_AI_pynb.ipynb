# **Calculadora Simples**

```python
# Solicita ao usuário o primeiro número e o converte para float
num1 = float(input("Digite o primeiro número: "))
# Solicita ao usuário o segundo número e o converte para float
num2 = float(input("Digite o segundo número: "))

# Solicita ao usuário a operação desejada (somar ou subtrair) e converte para minúsculas
operacao = input("Escolha a operação (somar/subtrair): ").lower()

# Verifica qual operação foi escolhida pelo usuário
if operacao == "somar":
    # Se a operação for 'somar', realiza a soma dos dois números
    resultado = num1 + num2
    # Imprime o resultado da soma
    print(f"A soma é: {resultado}")
elif operacao == "subtrair":
    # Se a operação for 'subtrair', realiza a subtração dos dois números
    resultado = num1 - num2
    # Imprime o resultado da subtração
    print(f"A subtração é: {resultado}")
else:
    # Se a operação não for 'somar' nem 'subtrair', imprime uma mensagem de erro
    print("Operação inválida. Por favor, escolha 'somar' ou 'subtrair'.")
```
